## 1. Introduction

### 1.1. The relevance of investigating dataset reuse

A lot of effort is invested into the sharing of data by researchers by creating infrastructures, developing know-how, and raising awareness. The practice of data sharing is called upon, supported, and sometimes required by funders \[@wellcome2025\], journals \[@loder2024; @plosmedicine\], institutions \[@theneuro2025\], scientific societies, and governments \[\] (UNESCO Recommendation on Open Science, 2023). Data sharing serves multiple purposes \[@gregory2020; @locher2023\]. The availability of data has a positive effect on trust \[@rosman2022\], and thus availability of data is a value in itself. However, we see a wide agreement that value is primarily created by data sharing when somebody else takes this data and uses it in some way. To quantify the impact of shared data, we thus focused on its reuse, and, more specifically, on references to dataset reuse in the published literature. Reuse can take place in a lot of ways, and by many actors: in addition to researchers, it could also be by governments, civil society, companies, or other actors. In addition, researchers might reuse data for purposes outside of research, as e.g. teaching \[@gregory2020\]. Thus, many reuse cases might never find their way into the literature. However, the traces of data reuse in the literature are currently the only way to investigate at scale, whether data is reused. Thus, we focus on mentions of data reuse in the published literature as a proxy of overall data reuse to investigate the impact of shared data. Ultimately, the answer to this question is central to advance and focus efforts regarding data sharing in the future \[@eosceden\].

### 1.2. Terminology

We are aware that data reuse is in itself not a self-evident concept, and it has been even argued that it cannot be defined at all \[@sandt2019\]. Although we argue that such a distinction is intuitive and useful, a line is certainly very difficult to draw. For the purpose of our analysis, we consider every instance where the article reporting data reuse (‘reusing article’) and the article reporting data sharing (‘sharing article’) do not have author overlap. Although this is certainly a simplification, it seems justified and our own analyses show that typically authors see this distinction the way we do, e.g. in that references to data reuse cases are much more often found outside of data availability statements than reference to what we define as ‘own data’. However, this issue warrants its own discussion, and probably a simple dichotomy might not suffice in the long term. In that regard, one of the authors (EB) is involved in the FORCE11 Working Group “Data Usage Typologies” \[@force11\], which aims to define types of data (re)use and their respective properties. **\[TODO: this section needs to be harmonized with the discussion\]**

The literature on data use and reuse employs different terms for statements which refer to utilized datasets. We follow the terminology employed by @gregory2023. Thus, we use the term ‘data citation’ for data objects in reference lists, while the term ‘data mention’ is used for statements in the publication text. As a consequence, the Data Citation Corpus analyzed here is, in this terminology, a corpus of both ‘data mentions’ and ‘data citations’ and, in a summarizing fashion, is denoted as a corpus of ‘data references’. ‘Data citations’ and ‘data mentions’ are synonymous with ‘formal data citations’ and ‘informal data citations’, respectively, as used by @park2018. The third type of data references which we analyze are ‘indirect data citations’ \[@sandt2019\]. In our case, this applies to cases where a data article rather than the dataset itself is referenced.

### 1.3. Institutional dataset reuse as specific application

We were interested in the reuse of data published by researchers from our institution. The primary focus of our study are the questions which datasets have been mentioned in the literature, and how often. In addition, we investigated the characteristics of reuse with regard to properties of data (e.g., whether it is data collected from human participants), properties of metadata (e.g., license), and the time lag between data publication and data reference.

We work at the Charité – Universitätsmedizin Berlin, a large university medical hospital, where very diverse biomedical research is undertaken. The Charité’s researchers published ca. 5600 reviewed articles in 2023, of which 79% are open access in some form. The research ranges from clinical and epidemiological studies to preclinical studies in animals, cell cultures, and in silico. A smaller subset of publications also stems from views onto medicine from other domains, including sociological, historical, and meta-research approaches. Information on publication access types, as well as several other metrics of open and responsible research can be found in an institutional Responsible Research Dashboard \[@charité\].

The list of datasets published by our institution is an outcome of the process of locally monitoring open science practices. We have collected information on the publication of datasets by Charité’s researchers, and thus, we know for the period 2020-2023 which datasets have been shared. This list can then be matched to references in the Data Citation Corpus to investigate the reuse of datasets from our institution. The process is briefly described in the methods section, which also provides references to detailed descriptions.

### 1.4. Data Citation Corpus as data source

The use of references to datasets is complicated by the fact that data reuse is rarely acknowledged by data citations \[@ninkov2021; @gregory2023\]. More commonly, data reuse is mentioned in a more or less informal way in different article sections \[@park2018; @sandt2019\]. ### 1.5. Indirect data citation via data articles

In addition to data citations and data mentions, many authors reference not the reused dataset directly, but an article which describes the results derived from the data, given that such an article exists \[@callaghan2012; @gregory2023; @yoon2019\]. Detecting the whole breadth of indirect data citations would require investigating every single article reference to articles by Charité authors to distinguish between indirect dataset citations and references to the content of the data article proper. This can currently not be reliably automated and is not feasible manually, as it would apply to an estimated high five-digit number of articles. We this focus on citations of dataset articles, where the probability of encountering indirect data citations can be expected to be substantially higher than amongst research articles generally. We have used dataset articles in two common data journals to track such indirect data citations. This complements the analysis of the DCC and allows for a more comprehensive picture of dataset reuse.

### 2.3. Detecting datasets published by our institution

#### 2.3.1. Detecting datasets via text mining of institutional articles with ODDPub

The process through which we obtained the list of shared datasets has been described in detail in our publications and @iarkaeva2024, as well as a detailed protocol \[@iarkaeva2022\]. Briefly, starting with an institutional bibliography, the steps are the following:

1.  Downloading of all institutional publication full-texts and conversion of PDFs to text files
2.  Screening of these texts with the ODDPub text mining tool \[@riedel; @nicoriedel2025\] to detect statements indicative of open data
3.  Verifying that the open data cases flagged by ODDPub are indeed open data cases by our definition
4.  As part of this verification process, we document the so-called “best identifier” of the dataset, e.g. preferring a DOI over an accession code, where both are available; however, some cleaning of this list was still necessary before a matching with the DCC could take place

#### 2.3.2. Using datasets detected with ODDPub as seed to detect additional datasets

In the detection procedure described in @iarkaeva2024, as well as briefly in section 2.3.1., we extracted one dataset identifier per repository. However, we had observed that in many cases multiple datasets had been deposed in the same repository, in some cases up to several dozen. Thus, we used the extracted dataset identifiers as “seeds” to detect further identifiers. These additional identifiers were only searched for either in the text surrounding the seed identifier, or if this text clearly indicated where in the text or supplements further identifiers were to be found.

#### 2.3.3. Detecting datasets via text mining using DataStet

DataStet is an alternative to ODDPub, which also screens article full-texts to detect dataset references **\[TODO datastet reference\]**. It includes some functionalities in addition to those offered by ODDPub, including a detection of whether a dataset had been at all created for the study, as well as a probability for the classification of a detected statement as indicative of shared data. However, we used DataStet only for the purpose of detecting any additional datasets which ODDPub might have missed. Unlike the datasets detected with ODDPub, the datasets detected using DataStet were not used as seeds for further dataset detection (see 2.3.2.).

#### 2.3.4. Indirect citation detection via data articles

The purpose of this analysis was to investigate for a sample of data articles by Charité authors, whether the articles citing the data articles referred to reuse of the presented data. In other words, we aimed at detecting indirect data citations.

Data in Brief and Scientific Data are, to our knowledge, the two most common data journals not specific to any discipline. All Charité articles published in these journals were detected on 16.8.2024 using publication data from Embase and Pubmed for 2016-2020 and openAlex for 2021 and 2022. For each data article, we determined whether the data presented was deposited in repositories, and if so, under which identifier(s). Of the 28 data articles, seven shared the data exclusively in the article itself or as supplementary materials. For these seven articles, no data identifiers were matched to the DCC. Of the remaining 20 articles, some had multiple datasets attributed to them, thus corresponding to 33 unique datasets overall (i.e., excluding deposits of the same dataset in two or more repositories).

Then, each article was searched in Google Scholar to obtain the list of articles which cited it, applying the “sort by date” function, thus using the most recent citations. For data articles which had up to 60 citations, all citing articles were investigated with regard to the reuse of data, with the goal of detecting possible indirect data citations. For articles with \>60 citations, due to time constraints caused by the substantial effort necessary for investigating the context of individual citations, only the first 60 eligible citing articles were checked. Articles were ineligible if it was not possible for us to access the citing article full-text and/or the language was not English. Out of 20 data articles, fogit addur have more than 60 eligible articles citing them on the respective date of checking.

For each citation of the respective data article within the citing article, it was checked, whether this was in fact an indirect data citation. All places where the data article was cited were investigated. We distinguished in a dichotomous fashion between cases of dataset reuse (i.e., indirect data citations), and citations which indeed referred to the data article itself. We did not distinguish between different purposes of data reuse (e.g., re-analysis, meta-analysis, comparison), or of data article references proper (e.g., supportive, contradictory, neutral).

Furthermore, depending on the journal it is harder to find where in article something was referenced. For example, in two articles which used numbers to indicate a citation, these numbers appeared in the document over a hundred times and therefore required too much time to find all citations of a Charité article.

### 2.4. Data cleaning and standardization

The list of datasets thus obtained had to be cleaned and standardized before matching with the DCC, in order to maximize the probability of matching. This processing included the following steps:

1.  removal of empty entries, non-identifier entries, and any free text beyond the identifier itself (i.e., DOI, link including an accession code, or link only)
2.  standardization of identifier notation; an automated function was used, which extracted the accession number part from each given URL of a dataset; e.g., removing “https://www.rcsb.org/structure/” from “https://www.rcsb.org/structure/5MEQ” to yield the accession code “5MEQ”; in addition, DOIs were standarized to a uniform format
3.  manual verification of the output to ensure that all datasets were indeed in the correct format (i.e., either a DOI or an accession code), with the version information removed
4.  deduplication (i.e., making sure the Charité dataset identifier list contains each identifier only once)

## 4. Discussion

### 4.1. Detecting data reuse through the DCC

Using two approaches, we charted the reuse of datasets shared alongside articles published by researchers from our large biomedical research institution. Our primary approach made use of the Data Citation Corpus (DCC) to detect references to reused datasets. Between 2020 and 2023, Charité researchers published (not considering the author position) check X articles for which at least one dataset was shared. This resulted in check \[AC\] datasets, which were referenced check X times. This is, to our knowledge, the first analysis to quantitatively show at the scale of a research institution, how the datasets produced are reused and enter the published literature. We show that reuse of datasets is common in the biomedical domain. At the same time, check \[AC\]% of datasets have not been referenced within the time period in question. While the distribution of dataset citations over time has to our knowledge so far not been investigated, a highly uneven distribution of dataset citations is plausible based on the literature on article citations \[@redner1998; @brzezinski2015\]. We did not investigate the fraction of dataset citations out of all dataset references to Charité articles. However, it has been shown repeatedly that the number of data citations alone is low. E.g., @ninkov2021 show that only ca. 1% of datasets in DataCite had a citation. This is corroborated by the survey of @gregory2023, which citing the article analyzing the data is much more common than citing the data itself.

It has been described extensively before \[e.g. @silvello2018\] that data citations can indicate both the sharing and the reuse of data. Correspondingly, we observed, in addition to check X cases of dataset reuse (also see below for a discussion of data sharing vs reuse), check X cases where authors mentioned datasets they had themselves created. The definition of “being created by Charité authors” used here is that a given dataset was mentioned in an article by Charité authors as underlying evidence they had themselves collected. This definition does not require overlapping authorship between the using article and the dataset itself. We do not know the first publication with which a dataset was shared, and thus we do not make a distinction between own sharing and own use – rather, we pool this in the data sharing category.  

On a methodological level, we also show that, given a list of dataset identifiers, this can be matched to the DCC to obtain a corresponding list of data citations. As we also describe, this process is far from straightforward, however. Even if the institutional data identifier list is highly standardized, the variability of dataset IDs in the DCC poses considerable challenges. The same dataset can be referred to in several ways, which makes the detection of references general-purpose repositories particularly complex. Nevertheless, the DCC includes both data citations and data mentions, and thus has a much broader coverage than databases which build on citation metadata only. This is necessarily true for DataCite \[analyzed e.g. by @ninkov2021\] which constitutes the subset of data citations within the DCC. It has been shown for datasets in the earth sciences, that proprietary databases, which include additional methods and sources, have better coverage \[@gerasimov2024\]. It is still to be investigated, how these databases compare to the DCC. \[check does Avihay have anything to add?\]

### 4.2. Detecting data reuse by indirect citations of data articles

We complemented this approach by an analysis of references to data articles. Data articles or ‘data descriptors’ are a type of article which explains in detail the provenance and properties of a dataset, but does not report results derived from its analysis. We examined a sample of 28 data articles published by Charité researchers between 2016 and 2022 in the two most common, non-disciplinary data journals. Our analysis has confirmed the finding of @jiao2020 that it is common practice in the literature to refer to a data article instead of the data, when actually the data have been used. At the same time, not every reference to a data article indicates a reuse of the data. @jiao2020 observed for two journals in the earth as well as physical and chemical sciences that 56% of citations to data articles indicate actual data reuse. In our case, 25.8% of citations to data articles (147 of 570) in fact referred to the underlying dataset (or one of several shared datasets). Thus, studies on the reuse of data need to take such “indirect citations” into account. These constitute a substantial fraction of all dataset references, and could be of particular importance for some purposes of monitoring, as data articles describe particularly ‘valuable’ datasets. Also, we see indications that this sample is characterized by both data types and types of data reuse distinct form the sample found via the DCC. However, this warrants further research.

To extract the information whether a data article reference in fact referred to the underlying data was manual and thus time-consuming. Correspondingly, we had to limit ourselves to extracting up to 60 citations per data article. In four out of 28 cases, the number of citations exceeded this number. In these cases, we have extrapolated the number of expected data uses based on the number of data uses within the 60 citations we analyzed. This extrapolation yielded a staggering number of 695 putative data reuse cases for the most-cited dataset. While there is an certainly an error margin attached to this number, it does clearly indicate the reuse value of certain well-curated and unique datasets.

### 4.3. Putative invisible data reuse

The numbers on data reuse reported here are probably very substantially underestimated. This is due to several reasons. For further limitations, which do not introduce an obvious directionality by decreasing detected reuse cases, see section 4.5.

1.  We extracted per article by Charité authors one dataset identifier per repository. However, especially for OMICS data, which are the most reused in our sample, it is common to share more than one dataset per repository. check E.g., ref had X datasets in GEO. In this case, the probability of detecting reuse is thus only check X%, as we extracted only one out of these check x identifiers (at least under the simplifying assumption that dataset citations are independent of each other). Thus, an unknown but substantial number of datasets is missing from our sample.

2.  The screen for shared datasets on which our analysis relies could not be performed on check ca. 20% (does Vladi have exact numbers) of all research articles published by our institution. These could not be accessed programmatically, either for licensing or for technical reasons.

3.  While some repositories create versioned identifiers or allow additions to a deposit under the same identifier, in other cases new versions of a dataset receive a fully new identifier. In such cases, detection by our method becomes impossible.

4.  We searched for references to Charité datasets using exact matching. Thus, we might have missed cases where the dataset references from DCC and the Charité screening pointed to the same dataset, but one referred to the DOI and the other to the accession code. This is possible where repositories as e.g. the Protein Structure Database (PDB) uses both accession codes and DOIs.

5.  The DCC serves to analyze references to datasets which have identifiers, either persistent identifiers (typically DOIs) or accession codes. However, restricted-access data sharing typically takes place via platforms which do not provide either. In our sample, the only two restricted-access platforms were dbGaP and EGA. Thus, references to restricted-access datasets especially outside of OMICS fields would not have been detected.

6.  We only extracted identifiers of shared datasets from the English-language literature.

7.  In the analysis of indirect references via data articles, we focused on the two most common, non-disciplinary data journals. However, over 30 journals, most of them disciplinary, exist, which offer data articles as a publication type (Data Journals, no date). check In addition, we set a cutoff at 60 references per data article, which was exceeded in **x** cases (relevant enough?)

For the reasons described above, we expect that the actual volume of dataset reuse as reported in the literature is substantially larger than we report. While no number can be attached to it, it is reasonable to assume that the overall number of references in the literature to datasets shared by Charité researchers is several times as large as the number we report.

We searched for reuse of datasets associated with Charité publications starting in 2020. Although some datasets were older, the majority was published in temporal proximity to the article where we detected it. Thus, at the time of the creation of the used DCC version two (check …..2024), most datasets were still comparatively new. The literature on data reuse \[@vries\] as well as our own analyses (see Fig. check xxx) indicate that dataset reuse accumulates over many years and at least for some time the number of citations per year increases. Thus, we would expect an overall number of citations to be substantially larger, if we had screened an earlier section of the Charité literature. Correspondingly, the datasets we analyzed can be expected to accumulate many more references in the future.

On a more general note, the approach we follow can only capture one type of data reuse, namely the reuse which contributes to published research and is considering worthy of mention by researcher. Data referencing practices differ between communities \[@gregory2023\], and it has been shown that not all data uses will be references in publication with equal probability (Borgman, 2016; Federer, 2019). For example, use for teaching or calibration might not commonly mentioned in research articles \[@gregory2020\]. In addition, the literature on publication bias (check ref) suggests that many data reuse cases might not be reported as the project they are part of is itself not published. Lastly, certain reuse purposes, in particular teaching, typically do not contribute to academic publishing, and thus cannot at all be detected.

### 4.5. What is data reuse anyway?

In this article, we have taken a pragmatic approach to data reuse. We have defined data reuse as the referencing in a research article of a dataset which is findable online, whether openly or with a defined access pathway, where the article referencing the data and the article for which we had originally detected data sharing had no author overlap. In our experience, it is a common view in the research community that a distinction between “own data use” and “data reuse” is possible. @gregory2023 similarly adopted the definition that “data reuse” means “using data which others have created, for any purpose.” The notion that use and reuse can be distinguished to us seems intuitive, as well as useful to inform ways to measure the value created by data sharing. This in turn is necessary to inform policy and support activities. However, this notion being intuitive and useful does not necessarily make it conceptually consistent. We would argue that it is intuitive based on common notions of “data ownership”, which might, however, be misleading. On the legal level, data cannot be owned at least in some jurisdictions, as e.g. in Germany \[@bmwk\]. Beyond that, starting from the related notion that data are intellectual property of some sort (whether this is legally enforceable or not), “data ownership” suggests a clear line between an “owner” or “creator” and a “non-owner” or “non-creator”. However useful such a distinction might be, defining it requires assumptions which are not necessarily in line with scientific practice, or at least not universally accepted. Consider the case real-world case of a database which makes datasets available, but requires that PIs from the original project are included as co-authors. By our definition, the resulting articles would be “own data use”. However, without data sharing, this article would not have existed, and thus value has been created by making data available to others. In turn, the notion that “reuse” is “use for a different purpose”, independently of authorship, is also widely held and so intuitive that @gregory2023, despite the definition given above, also state that “roughly three-quarters of respondents \[…\] reported reusing their own data multiple times”. It has been argued by @sandt2019 that reuse cannot at all be defined. While we consider our definition legitimate and useful for our purposes, there is need for a community consensus on what constitutes data reuse. The FORCE11 Working Group “Data usage typologies” \[@force11\] will work on this question between early 2025 and mid 2026. In this regard, it is important to mention that data references are indicative of data use, but similar to views or downloads of datasets, they do not constitute data use in themselves. For the purpose of this work we also have to stress that our definition of own data sharing vs. data reuse creates borderline cases in both directions. In some cases, publications using data from Charité researchers (or their collaborators) were excluded from the analysis, if there was author overlap, as this is not “data reuse” by our definition. From the type of data and the authorship position we would however expect that the collaboration came about precisely by way of the data being reused. Vice versa, we also observed cases where we included datasets into our sample as “Charité datasets”, although the creation date of the original dataset, and the link between its study PIs and the article with Charité participation that the dataset had actually been reused. It is also important to mention that we understand “data” throughout this article as a shorthand for “datasets”. It is datasets which are shared and datasets which are reused. Any piece of information which can be used to support a claim can be understood as “data”, which would not be a useful category. In line with this understand we have excluded cases of databases which contain information as e.g. the association between a certain disease and a certain mutation, but not a dataset in the sense of structured information which can be downloaded and processed further. Necessarily, the distinction between data and datasets is a matter of definition and contentious, as is the related distinction between data and metadata.

### 4.6. Limitations

The following restrictions and limitations apply to our analysis:

· We only detect reuse detected through references in the published literature. This excludes many types of data reuse, as described in… \[I think we have it somewhere in the text already\]. In addition, dataset reuse by commercial actors is common and cannot be detected via bibliographic methods \[@ross2024\].

· While we have monitored the sharing of datasets starting with publications from 2016, we have documented the dataset identifier in a consistent way starting with publications from 2020. Thus, we analyze reuse of datasets found through publications from the period 2020-2022. The datasets themselves could have been published earlier, however.

· Over the years, our criteria for inclusion have somewhat changed. For instance, we decided at some point to include a “persistence” requirement, which leads e.g. to the exclusion of deposits on Github. However, given the reuse cases we detected up to now (mostly accession codes in disciplinary repositories), it seems probable that this drift over time did not influence our analysis as described here.

· When we speak of datasets “from our institution”, we refer to datasets underlying articles with at least one co-author from the Charité. This does not imply that the dataset was actually generated or shared by Charité authors. This would be impossible to know from the literature alone.

· The article we determined as sharing data doesn’t have to be the first one to have shared the data, and this is indeed what we observed. We detected \*one\* paper from the Charité which shared the data, but it doesn’t have to be the first one.

· Our definition disregards the question of dataset authorship, and we analyze article authorship only. Unfortunately, not all repositories provide authorship information, but where it is available it would be possible to define reuse as a lack of authorship overlap between an article and a dataset. This could be more reflective of actual data sharing and reuse processes, but whether it really is would warrant further investigation.

· In our analysis of the DCC, we did not consider the purposes or context of dataset references. We assume that datasets are typically referenced when data have been used. However, this must not be the case \[@gregory2023\], and it is well possible that data can be referenced without using them, e.g. to legitimize own work or using the dataset as a proxy or additional support for scientific claims made in research articles. Our analysis of references to data articles indicates that at least if datasets are described in detail, check a majority (is that the case?) of references does not indicate reuse of the data, and this support the assumption that purposes of referencing data directly are also very diverse.

# Methods

Identifiers and accession codes of open an restricted datasets that were found in articles by Charité’s authors between the years 2020 and 2023 (henceforth referred to as the datasets) were collected as part of a separate data sharing monitoring process. Articles were defined as written by Charité’s authors if at least one author had an affiliation to Charité.

Prior to looking for mentions of Charités datasets in the DCC, the datasets were standardized into a format that would maximize their chance to be found in the DCC. This was done by an automatic standardization function that extracted the accession number from each given URL of a dataset, and standardized any DOI to a uniform format. The output of that function was verified manualy in order to ensure that all datasets are indeed in the correct format (i.e. either a DOI or an accession code), with the version information removed from it.

Prior to applying several preprocessing and cleaning steps to the DCC as well, we first explored the corpus in order to make sure that there are no NAs in relevant fields. We then made sure that dates entries of pucblication years were correct. Lastly, we observed the frequency of repositories in which datasets were deposited according to the corpus, together with publication years of the mentioning articles. Once we made sure that the fundamental structure of the corpus is preoperly defined, we went on with the preprocessing steps described below.

DOIs of datasets deposited in general repositories (i.e., Zenodo, Figshare, Dryad, and OSF) were found to appear in multiple formats within the DCC. To ensure consistency, all DOIs were standardized to a uniform format. As with the Charité datasets, version information was removed from all entries. Subsequently, excess characters were stripped from both the dataset and the articles’ DOIs. This included leading and trailing slashes, dots and spaces, together with any other space found within the string value of the dataset or the article’s DOI. Additionally, commas were replaced with dots. DOIs of articles were likewise standardized to a uniform DOI format.

The DCC had also cases where the same DOI was listed both in the “mentioning article” field and the “mentioned dataset” field. There were cases where this was discovered only after standardizing both fields. All such cases were removed from the DCC. We have also decided to not include preprints in the analysis, and therefore all cases where the repository was listed as a preprint service were removed as well.

Following all the steps above, we looked for Charité’s datasets mentions in the DCC. This step resulted in a list with cases from the DCC which had a matching dataset identifier to the ones in the Charité datasets list. From this results list, cases where the mentioning article and the article with the mentioned dataset had at least one author in common were excluded. For this step, meta data regarding the authors of each article was extracted using Openalex. Publication years of Charité’s papers and of their datasets were incorporated into the results list as meta data as well.

Lastly, a sample of the remaining cases was taken in order to manualy verify the results, focusing on the following cases:

1\. Mentioning article is listed as published before 2016

2\. Mentioning article is listed as published before the publication of the dataset

3\. Publication year of mentioning article is listed as earlier than publication year of the Charité article.

venn_ellipse_plot <- function(setlist, a = 1.6, b = 1.0, colors = c("#E6E6E6", "#A6CEE3", "#B2DF8A")) {
  stopifnot(length(setlist) == 3)
  
  # Control circle's size and shape
  ellipses <- tibble(
    group = names(setlist),
    x0 = c(1, 2, 1.5),
    y0 = c(1, 1, 2),
    a  = a,
    b  = b,
    angle = 0
  )
  
  names(colors) <- names(setlist)
  
  # Extract sets
  A <- unique(setlist[[1]])
  B <- unique(setlist[[2]])
  C <- unique(setlist[[3]])
  
  total <- length(unique(c(A, B, C)))
  
  # Define Venn regions
  regions <- tibble(
    label = c(
      "A only", "B only", "C only",
      "A & B", "A & C", "B & C", "A & B & C"
    ),
    count = c(
      length(setdiff(A, union(B, C))),
      length(setdiff(B, union(A, C))),
      length(setdiff(C, union(A, B))),
      length(intersect(A, B)) - length(Reduce(intersect, list(A, B, C))),
      length(intersect(A, C)) - length(Reduce(intersect, list(A, B, C))),
      length(intersect(B, C)) - length(Reduce(intersect, list(A, B, C))),
      length(Reduce(intersect, list(A, B, C)))
    )
  ) |>
    mutate(
      pct = count / total,
      label_text = sprintf("%d\n(%.1f%%)", count, 100 * pct),
      
      # control text labels locations
      x = c(0.4, # right
            2.6, # left
            1.5, # top
            1.5, # bottom inter
            0.8, # left inter
            2.3, # right inter
            1.5), # cented inner
      y = c(1,
            1,
            2.5,
            0.7, 
            1.75,
            1.75,
            1.5)
    )
  
  # control titles locations
  titles <- tibble(
    group = names(setlist),
    x = c(0.6, 2.4, 1.5),
    y = c(-0.2, -0.2, 3.2)  # top/top/bottom
  )
  
  ggplot() +
    geom_ellipse(
      data = ellipses,
      aes(x0 = x0, y0 = y0, a = a, b = b, angle = angle, fill = group),
      color = "black",
      linewidth = 0.7,
      alpha = 0.4
    ) +
    geom_text(
      data = regions,
      aes(x = x, y = y, label = label_text),
      size = 5
    ) +
    geom_text(
      data = titles,
      aes(x = x, y = y, label = group),
      size = 7
    ) +
    scale_fill_manual(values = colors) +
    coord_fixed() +
    theme_void() +
    theme(legend.position = "none") # comment out in order to include legends
}
